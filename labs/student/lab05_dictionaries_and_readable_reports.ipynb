In Lab 04, `("P1", 80, 28)` meant label, battery percentage, and voltage. You had to remember the order. A dictionary stores values under keys:

| Syntax | Meaning |
|---|---|

| `record["voltage_v"]` | Read the value under that key |

### A2. Pass named records through a batch function | 15 minutes

This complete classifier uses the same four rules as Lab 04. It is supplied and correct. Leave it unchanged.

1. CRITICAL if battery is below 15% or voltage is below 22 V.
2. SAFE MODE if battery is below 25% or voltage is below 24 V.
3. WARNING if battery is at most 35%.
4. NOMINAL otherwise.

All measurements here are valid numbers. The new part is looking up the two inputs by their keys. A list holds the records in observation order; each dictionary holds the fields within one record.


````python
def classify_power(battery_percent, voltage_v):
    if battery_percent < 15 or voltage_v < 22:
        return "CRITICAL"
    elif battery_percent < 25 or voltage_v < 24:
        return "SAFE MODE"
    elif battery_percent <= 35:
        return "WARNING"
    else:
        return "NOMINAL"

power_records = [
    {"label": "P1", "battery_percent": 80, "voltage_v": 28},
    {"label": "P2", "battery_percent": 10, "voltage_v": 28},
    {"label": "P3", "battery_percent": 80, "voltage_v": 21},
    {"label": "P4", "battery_percent": 15, "voltage_v": 28},
    {"label": "P5", "battery_percent": 35, "voltage_v": 28},
    {"label": "P1", "battery_percent": 80, "voltage_v": 28},
]
````

The batch function below has one deliberate mistake: it looks up the wrong field in one place. Before running, inspect both arguments passed to `classify_power`.

Repair the mistaken lookup. Keep the rest of the function's behavior: call the classifier once per record, return a new list of named results in the input order, keep repeated labels, and return `[]` for no records. Do not change the input records or print inside the function.

`results.append({"label": ..., "status": ...})` creates a new result dictionary and adds it to the result list. In the display loop, `result` is one of those dictionaries, not the whole list.


````python
def classify_named_batch(records):
    results = []
    for record in records:
        status = classify_power(record["battery_percent"], record["battery_percent"])
        results.append({"label": record["label"], "status": status})
    return results

named_results = classify_named_batch(power_records)
for result in named_results:
    print(result["label"], result["status"])
````
